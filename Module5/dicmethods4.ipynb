Dictionary Method

```python
dic = {"name": "Alice", "age": 30, "city": "New York" , "numbers": [1, 2, 3, 4, 5], "age": 45}

print(dic)
```

```python
print (dic.get("new"))
print (dic.get("new", "Not Found"))
```

```python
# adding a value

dic = {"name": "Alice", "age": 30, "city": "New York" , "numbers": [1, 2, 3, 4, 5], "age": 45}
dic["email"] = "alice@example.com"
print(dic)

dic.update({"money": 1000})
print(dic)
```

```python
# deleting

del dic ["email"]
print(dic)
```

```python
# copy

new_dic = dic.copy()
print(new_dic)
```

```python
# testing

# dic_3 = {[10, 20, 30]: "A"}.........not allowed as list is mutable
dic_3 = {(10, 20, 30): "A"} # allowed as tuple is immutable
print(dic_3)
```
