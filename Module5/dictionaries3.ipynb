```python
# empty
dic = {}
print(type(dic))

dic = {"name": "Alice", "age": 30, "city": "New York" , "numbers": [1, 2, 3, 4, 5]}
print(type(dic), dic)
```

```python
# access
# dic_name[key]

print(dic.get("name"))
print(dic["name"])
print(dic["age"])
print(dic["city"])
print(dic["numbers"])

# change value
dic["name"] = "Bob"
print(dic)

# we can't keep 2 value for the same key, it will overwrite the previous value
dic = {"name": "Alice", "age": 30, "city": "New York" , "numbers": [1, 2, 3, 4, 5], "age": 45}
print(dic)
```
