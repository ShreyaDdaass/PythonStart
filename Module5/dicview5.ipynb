Dictionary View Method

```python
dic = {"name": "Alice", "age": 30, "city": "New York" , "numbers": [1, 2, 3, 4, 5]}

keys = dic.keys()
print(keys)

dic ["country"] = "USA"
print(keys)

values = dic.values()
print(values)

dic ["age"] = 50
print(values)

# in items things will come as tuples ...first brackets
items = dic.items()
print(items)
```

Iteration

```python
for key, value in dic.items():
    print(key,value)
```
