```python
# Declaration

tup =(10, 20, 30, 40)
float_tup = (10.2, 3.54, 95.44)
print (type(float_tup), float_tup)


mixed_tup = (10, 10.3, 65, 992.366, "str", True)
# kpokrfhorei (this type wont be allowed)
print (mixed_tup)

lst = [10,20,30]
tup1 = tuple([lst])
print(type(tup1), tup1)
```

```python
# access 
tup = (10,20,30,40,50.3, "str", False)

# index
print(tup[1])

# slicing
new_tup = tup[0:3]
print (new_tup)

```

```python
# immutability

lst = [10,20,30,40]
lst.append(89)
lst[1] = 566
print(lst)
tup = (10,20,30,40)
tup[1] = 55
print(tup)
```
