```python
N = int(input())
target = float(input())

total = 0

for _ in range(N):
    loss = float,(input().split())
    total += loss

average = total / N

if average <= target:
    print("PASS")
else:
    print("RETRY")
```

```python
N = int(input())

yes_count = 0
no_count = 0

for _ in range(N):
    vote = input()
    
    if vote == "YES":
        yes_count += 1
    else:
        no_count += 1

if yes_count >= no_count:
    print("ACCEPT")
else:
    print("REJECT")
```

```python
message = input()

if "happy" in message or "joy" in message or "smile" in message:
    print("Happy Mood")
elif "sad" in message or "cry" in message or "angry" in message:
    print("Sad Mood")
else :
    print ("Neutral Mood")
```

```python
pixels = list(map(int, input().split()))

average = sum(pixels) / len(pixels)

if average < 85:
    print("Dark Image")
elif average <= 170:
    print("Normal Image")
else:
    print("Bright Image")
```
